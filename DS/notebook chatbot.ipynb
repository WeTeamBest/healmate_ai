## Cek Anomali "utterance"

```python
# Kolom Sumber
col = "utterance"
```

```python
# Pattern
pattern = r"hit:\d+_conv:\d+.*?(?=hit:\d+_conv:|$)"
```

```python
# List Final
final_rows = []
```

```python
# Loop Semua Baris
for _, row in ed_gabungan.iterrows():

    text = str(row[col])

    # cari semua looping data
    matches = re.findall(pattern, text, flags=re.DOTALL)

    # SIMPAN BARIS ASLI

    cleaned_text = re.sub(
        pattern,
        "",
        text,
        flags=re.DOTALL
    ).strip()

    original_row = row.to_dict()
    original_row[col] = cleaned_text

    final_rows.append(original_row)

    # SPLIT SEMUA hit:_conv:

    for match in matches:

        # split berdasarkan koma
        parts = [x.strip() for x in match.split(",")]

        # pastikan minimal 8 kolom
        while len(parts) < 8:
            parts.append(None)

        # PERTAHANKAN FORMAT conv_id ASLI

        conv_match = re.search(
            r"(hit:\d+_conv:\d+)",
            parts[0]
        )

        conv_id = conv_match.group(1) if conv_match else parts[0]

        # ROW BARU

        new_row = {
            "conv_id": conv_id,
            "utterance_idx": parts[1],
            "context": parts[2],
            "prompt": parts[3],
            "speaker_idx": parts[4],
            "utterance": parts[5],
            "selfeval": parts[6],
            "Unnamed: 8": parts[7]
        }

        final_rows.append(new_row)
```

```python
# Dataframe Final
data_final = pd.DataFrame(final_rows)
```

```python
# List Hasil
chat_rows = []
```

```python
# GROUP PER CONVERSATION

for conv_id, group in ed_final.groupby("conv_id"):

    group = group.reset_index(drop=True)

    emotion = group["emotion_group"].iloc[0]

    # PAIRING USER -> BOT
    # speaker_idx:
    # 0 = user
    # 1 = bot

    for i in range(len(group) - 1):

        current_speaker = str(group.loc[i, "speaker_idx"])
        next_speaker = str(group.loc[i + 1, "speaker_idx"])

        # ambil pasangan user -> bot
        if current_speaker == "0" and next_speaker == "1":

            input_text = group.loc[i, "utterance_clean"]
            label_text = group.loc[i + 1, "utterance_clean"]

            # skip kosong
            if pd.isna(input_text) or pd.isna(label_text):
                continue

            if str(input_text).strip() == "":
                continue

            if str(label_text).strip() == "":
                continue

            chat_rows.append({
                "conv_id": conv_id,
                "input": input_text,
                "label": label_text,
                "emotion": emotion
            })
```

```python
# 1. Emotion lexicon
#    TANPA core & secondary

emotion_dict = (
    emotion_words
    .groupby("emotion")["word"]
    .apply(list)
    .to_dict()
)

# 2. Preprocessing

def preprocess(text):

    text = str(text).lower()

    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text.split()

# 3. Emotion scoring

def calculate_emotion_probabilities(text):

    tokens = preprocess(text)

    token_counts = Counter(tokens)

    scores = {
        "anger": 0,
        "anxiety": 0,
        "acceptance": 0
    }

    # hitung frekuensi kata emosi

    for emotion, vocab_list in emotion_dict.items():

        for word in vocab_list:

            if word in token_counts:

                scores[emotion] += token_counts[word]

    # acceptance diturunkan jika anger/anxiety tinggi

    negative_strength = (
        scores["anger"] +
        scores["anxiety"]
    )

    if negative_strength > 0:

        scores["acceptance"] *= 0.35

    # smoothing kecil

    for emotion in scores:
        scores[emotion] += 0.01

    # softmax normalization

    exp_scores = {
        emotion: np.exp(score)
        for emotion, score in scores.items()
    }

    total = sum(exp_scores.values())

    probs = {
        f"{emotion}_prob": (
            exp_scores[emotion] / total
        )
        for emotion in exp_scores
    }

    return pd.Series(probs)

# 4. Apply

emotion_probs = ec_final["input_clean"].apply(
    calculate_emotion_probabilities
)

ec = pd.concat(
    [
        ec_final[
            [
                "input_clean",
                "label_clean"
            ]
        ].reset_index(drop=True),

        emotion_probs.reset_index(drop=True)
    ],
    axis=1
)

# 5. Predicted emotion

emotion_cols = [
    "anger_prob",
    "anxiety_prob",
    "acceptance_prob"
]

ec["predicted_emotion"] = (
    ec[emotion_cols]
    .idxmax(axis=1)
    .str.replace("_prob", "", regex=False)
)

# 6. Confidence

ec["emotion_confidence"] = (
    ec[emotion_cols]
    .max(axis=1)
)

# 7. Final output

ec = ec[
    [
        "input_clean",
        "label_clean",
        "anger_prob",
        "anxiety_prob",
        "acceptance_prob",
        "predicted_emotion",
        "emotion_confidence"
    ]
]

print(ec.head())
```

```python
# 1. Dictionary contractions

contractions_dict = {

    # basic
    "i'm": "i am",
    "i've": "i have",
    "i'd": "i would",
    "i'll": "i will",

    "you're": "you are",
    "you've": "you have",
    "you'd": "you would",
    "you'll": "you will",

    "we're": "we are",
    "we've": "we have",
    "we'd": "we would",
    "we'll": "we will",

    "they're": "they are",
    "they've": "they have",
    "they'd": "they would",
    "they'll": "they will",

    "he's": "he is",
    "he'd": "he would",
    "he'll": "he will",

    "she's": "she is",
    "she'd": "she would",
    "she'll": "she will",

    "it's": "it is",
    "that's": "that is",
    "there's": "there is",
    "what's": "what is",
    "who's": "who is",

    # negatives
    "don't": "do not",
    "doesn't": "does not",
    "didn't": "did not",

    "can't": "cannot",
    "couldn't": "could not",
    "won't": "will not",
    "wouldn't": "would not",
    "shouldn't": "should not",

    "isn't": "is not",
    "aren't": "are not",
    "wasn't": "was not",
    "weren't": "were not",

    "haven't": "have not",
    "hasn't": "has not",
    "hadn't": "had not",

    # misc
    "let's": "let us",
    "ma'am": "madam",
    "y'all": "you all"
}

# 2. Function cleaning text

def clean_text(text):

    text = str(text).lower()

    # expand contractions
    for contraction, expanded in contractions_dict.items():

        pattern = r'\b' + re.escape(contraction) + r'\b'

        text = re.sub(
            pattern,
            expanded,
            text
        )

    # hapus spasi sebelum tanda baca

    text = re.sub(r'\s+([.,!?])', r'\1', text)

    # rapihin multiple spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# 3. Apply cleaning

ec["input_clean"] = ec["input_clean"].apply(
    clean_text
)

ec["label_clean"] = ec["label_clean"].apply(
    clean_text
)

# 4. Preview hasil

print(
    ec[
        [
            "input_clean",
            "label_clean"
        ]
    ].head()
)
```
